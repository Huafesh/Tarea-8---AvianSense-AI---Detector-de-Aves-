# Clasificación de Aves con TensorFlow (Keras)
Este cuaderno cumple con los requerimientos para cargar un modelo `.h5`, preprocesar la imagen y realizar una predicción sobre una imagen de un ave.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files

print("TensorFlow version:", tf.__version__)
```

## 1. Cargar el Modelo .h5
Se utiliza `files.upload()` para poder subir el modelo directamente a Google Colab sin depender de rutas en Drive.

```python
# Sube aquí tu archivo .h5 generado en Teachable Machine
print("Por favor, sube tu archivo del modelo (ej. keras_model.h5):")
uploaded_model = files.upload()
model_filename = list(uploaded_model.keys())[0]

# Cargar el modelo
model = tf.keras.models.load_model(model_filename, compile=False)
print("\n¡Modelo cargado exitosamente!")
```

## 2. Cargar y Mostrar Imagen de Prueba (PIL y Matplotlib)

```python
print("Por favor, sube una imagen de un ave para analizar:")
uploaded_img = files.upload()
image_filename = list(uploaded_img.keys())[0]

# Usar PIL para abrir la imagen
img = Image.open(image_filename).convert('RGB')

# Usar matplotlib para mostrar en pantalla
plt.imshow(img)
plt.axis('off')
plt.title('Imagen de prueba subida')
plt.show()
```

## 3. Preprocesamiento y Predicción
Aplicamos exactamente el mismo procesamiento que en la interfaz web (JS): 224x224, expandir dimensiones y normalizar / 255.0.

```python
# 1. Cambiarle el tamaño a 224x224
img_resized = img.resize((224, 224))

# 2. Convertir imagen a array de numpy
img_array = np.array(img_resized)

# 3. Normalizar los píxeles (dividiendo entre 255.0)
img_normalized = img_array / 255.0

# 4. Expandir dimensiones: (224, 224, 3) -> (1, 224, 224, 3)
img_ready = np.expand_dims(img_normalized, axis=0)

# 5. Ejecutar model.predict()
predictions = model.predict(img_ready)

print("\n--- Resultados de Predicción ---")
print("Array crudo de predicciones:", predictions)
predicted_class = np.argmax(predictions)
confidence = np.max(predictions) * 100
print(f"\nLa IA determina que la imagen pertenece a la Clase (índice) {predicted_class}")
print(f"Nivel de confianza: {confidence:.2f}%")
```
